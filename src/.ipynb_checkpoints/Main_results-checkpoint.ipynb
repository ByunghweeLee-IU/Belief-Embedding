# Exploring the Belief Embedding Model

- In this notebook, we provide a simplified implementation for reproducing the key experimental results of the paper: **"Neural Embedding of Beliefs Reveals the Role of Relative Dissonance in Human Decision-Making"** by Lee et al.

- We include a method to download a fine-tuned S-BERT model via Hugging Face.
Prior to this code, we fine-tuned a pre-trained S-BERT model (roberta-base-nli-stsb-mean-tokens) using a training set composed of belief triplets, optimized with the triplet loss function. Each belief triplet consists of an anchor belief, a positive belief (contextually aligned with the anchor), and a negative belief (contextually distant from the anchor).


- For details on the methods used to generate these belief triplets, please refer to the main paper and '01_Data_generation_and_model_finetuning.ipynb' for detailed implementation.

- In the main manuscript, we trained five different models using a 5-fold train/test split for cross-validation of our experimental results. However, for simplicity and to obtain results more quickly, we use only a single model (the first of the five) in this notebook.

## Downloading the fine-tuned S-BERT model 

```python
import os
import numpy as np
import torch
from sentence_transformers import SentenceTransformer
```

```python
#download model using SentenceTransformer library from HuggingFace
#fine-tuned model 
model = SentenceTransformer("Byunghwee/roberta_belief_finetuned")

#base S-BERT model (before fine-tuning) for comparison 
basemodel = SentenceTransformer('roberta-base-nli-stsb-mean-tokens')        
```

## Obtaining vector representations of beliefs

* to obtain a vector representation (768 dim) of a belief statement:
 - In our study, a belief statement for a debate title is constructed by combining a template phrase ("I agree (disagree) with the following"), representing a voter's stance (Pro/Con), with the debate title, forming the statement: "I agree with the following: {Debate Title}."
 - Consider an example of three belief statements:

```python
sample_belief1 = "I agree with the following: Abortion should be legal."
sample_belief2 = "I agree with the following: Abortion is morally justified"
sample_belief3 = "I disagree with the following: Abortion"

sample_vector1 = model.encode(sample_belief1)
sample_vector2 = model.encode(sample_belief2)
sample_vector3 = model.encode(sample_belief3)

#let's check base S-BERT model too! 
sample_vector1_base = basemodel.encode(sample_belief1)
sample_vector2_base = basemodel.encode(sample_belief2)
sample_vector3_base = basemodel.encode(sample_belief3)
```

 - Belief 1 and Belief 2 share a similar stance (pro-abortion), while Belief 3 (anti-abortion) holds the opposite stance. Let's check if the first two beliefs are close in the belief embedding space!


```python
from sklearn.metrics.pairwise import cosine_distances, euclidean_distances

# pair wise distance between beliefs 
cosine_dist_matrix = cosine_distances([sample_vector1, sample_vector2, sample_vector3])
euclidean_dist_matrix = euclidean_distances([sample_vector1, sample_vector2, sample_vector3])

euclidean_dist_matrix_base = euclidean_distances([sample_vector1_base, sample_vector2_base, sample_vector3_base])

# output results 
print('Pairwise distance (Fine-tuned model)')
print(euclidean_dist_matrix, '\n')

print('Pairwise distance (base model)')
print(euclidean_dist_matrix_base)
```

```python
from itertools import combinations

print('Pairwise distance (Fine-tuned model)')
for i,j in combinations(range(3), 2):
    print(f"Distance between Belief {i+1} and Belief {j+1}: {euclidean_dist_matrix[i, j]:.2f}")

print()
print('Pairwise distance (base model)')
for i,j in combinations(range(3), 2):
    print(f"Distance between Belief {i+1} and Belief {j+1}: {euclidean_dist_matrix_base[i, j]:.2f}")
```

Both the fine-tuned and base models capture the relationship between beliefs, where Beliefs 1 and 2 are close, while Belief 3 is relatively distant. The fine-tuned model exhibits a clearer separation of these beliefs compared to the base model.

**Let's try another example**. 
 
 * Previously we have tested beliefs under a same topic (abortion). This time we add another type of belief, belief in God -- "I agree with the following: God exists"  

```python
sample_belief4 = "I agree with the following: God exists." 
sample_vector4 = model.encode(sample_belief4)
sample_vector4_base = basemodel.encode(sample_belief4)

sample_belief5 = "I disagree with the following: God exists." 
sample_vector5 = model.encode(sample_belief5)
sample_vector5_base = basemodel.encode(sample_belief5)
```

```python
euclidean_dist_matrix2 = euclidean_distances([sample_vector1, sample_vector2, sample_vector3, sample_vector4, sample_vector5])
np.fill_diagonal(euclidean_dist_matrix2, np.inf)

euclidean_dist_matrix2_base = euclidean_distances([sample_vector1_base, sample_vector2_base, sample_vector3_base, sample_vector4_base, sample_vector5_base])
np.fill_diagonal(euclidean_dist_matrix2_base, np.inf)
```

```python
#sample_belief1 = "I agree with the following: Abortion should be legal."
#sample_belief2 = "I agree with the following: Abortion is morally justified."
#sample_belief3 = "I disagree with the following: Abortion"
#sample_belief4 = "I agree with the following: God exists." 

simplified_beliefs = ["'PRO, Abortion should be legal.'", "'PRO, Abortion is morally justified.'", "'CON, Abortion'", "'PRO, God exists'", "'CON, God exists'" ] 

print('Pairwise distance (Fine-tuned model)')
for i in range(5):
    if not i==3:
        print(f"Distance between Belief4 {simplified_beliefs[3]} and Belief{i+1} {simplified_beliefs[i]} : {euclidean_dist_matrix2[3, i]:.2f}")



print()
print('Pairwise distance (base model)')
for i in range(5):
    if not i==3:
        print(f"Distance between Belief4 {simplified_beliefs[3]} and Belief{i+1} {simplified_beliefs[i]} : {euclidean_dist_matrix2_base[3, i]:.2f}")


print("\n * 'PRO/CON represents the phrase 'I agree / disagree with the following:'")
```

 * In the fine-tuned model, the distance between 'PRO, God exists' and 'CON, Abortion' is smaller than that between 'PRO, God exists' and 'CON, God exists'. However, in the base model, 'PRO, God exists' is closer to 'CON, God exists' than to 'CON, Abortion'.
 * This suggests that the fine-tuned model better captures real-world ideological alignments, whereas the base model relies more on surface-level lexical similarities.

# Evaluation of the fine-tuned model 

## Triplet evaluation

 * In the example above, we verified that the model captures semantic and contextual similarities between beliefs.
 * In this section, we systematically evaluate the model's performance using the triplet evaluation score, leveraging belief triplets generated from Debate.org data (training and test sets). We sample 10,000 triplets from both training and test dataset.

```python
#load triplets
import pickle 

train_path = '../dataset-robust/full_data/train_triplet_idx0.p'
test_path = '../dataset-robust/full_data/test_triplet_idx0.p'
        
with open(train_path,'rb') as f:
    train_data = pickle.load(f)

with open(test_path,'rb') as f:
    test_data = pickle.load(f)
```

```python
print("Original number of triplets (Train/Test)", len(train_data), len(test_data))
```

```python
#sampling 10000 triplets 
import pandas as pd
from tqdm import tqdm
from sentence_transformers import InputExample

train_data = pd.Series(train_data)
train_data = train_data.sample(10000, random_state=42).to_list()

train_samples = []
for e in tqdm(train_data):
    train_samples.append(InputExample(texts = e))

test_data = pd.Series(test_data)
test_data = test_data.sample(10000, random_state=42).to_list()

test_samples = []
for e in tqdm(test_data):
    test_samples.append(InputExample(texts = e))
```

```python
from sentence_transformers.evaluation import TripletEvaluator

#evaluate training data
triplet_evaluator = TripletEvaluator.from_input_examples(
                train_samples,
                write_csv=True,
                show_progress_bar=True,
                name = "sample_triplet_eval-train")

os.makedirs('../evaluation/', exist_ok=True)
triplet_evaluator(model, output_path='../evaluation/')
```

```python
#evaluate test data
triplet_evaluator = TripletEvaluator.from_input_examples(
                test_samples,
                write_csv=True,
                show_progress_bar=True,
                name = "sample_triplet_eval-test")

os.makedirs('../evaluation/', exist_ok=True)
triplet_evaluator(model, output_path='../evaluation/')
```

 * Let's compare the **triplet evaluation scores**

```python
train_score = pd.read_csv('../evaluation/triplet_evaluation_sample_triplet_eval-train_results.csv')
test_score = pd.read_csv('../evaluation/triplet_evaluation_sample_triplet_eval-test_results.csv')

print("===Training set score===") 
print(train_score[['accuracy_cosinus', 'accuracy_euclidean']].head(), '\n')

print("===Test set score===") 
print(test_score[['accuracy_cosinus', 'accuracy_euclidean']].head())
```

 * The results above show that the fine-tuned S-BERT model can classify belief relationships with an accuracy of up to **94.6% for triplets sampled from the training set** and **68% for unseen triplets from the test set**.

### PCA of belief vectors

```python
#load belief vectors 

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/belief_embedding_trainset_idx0.p', 'rb') as f:
    belief_embedding = pickle.load(f) #A dictionary where belief statements are keys and embedding vectors are values
```

```python
belief_keys = list(belief_embedding.keys())
belief_vectors = list(belief_embedding.values())

idx2belief = {}
belief2idx = {}

for i in range(len(belief_keys)):
    idx2belief[i] = belief_keys[i]
    belief2idx[belief_keys[i]] = i
```

```python
from sklearn.decomposition import PCA

pca = PCA(random_state=42)
pca.fit(belief_vectors)
pca_vectors = pca.transform(belief_vectors)

pca_x_vals = np.array([v[0] for v in pca_vectors])
pca_y_vals = np.array([v[1] for v in pca_vectors])
dat_pc = pd.DataFrame({'x':pca_x_vals, 'y':pca_y_vals}) #PC1-PC2 
```

#### Representing beliefs related to specific keywords

```python
def get_related_beliefs(keywords, pc_x=0, pc_y=1): 
    #pc_x, pc_y = pc index return related beleifs 
    related_beliefs = set()
    related_beliefs_vectors = []
    related_beliefs_pcvec = []

    for e in belief_keys:
        for k in keywords:
            if k.lower() in e.lower(): #convert to lower case 
                related_beliefs.add(e)
    
    related_beliefs = list(related_beliefs)
    for b in related_beliefs:
        b_idx = belief2idx[b]
        related_beliefs_vectors.append(belief_vectors[b_idx])
        related_beliefs_pcvec.append(pca_vectors[b_idx])

    related_beliefs_vectors = np.array(related_beliefs_vectors)            
    related_beliefs_pcvec   = np.array(related_beliefs_pcvec)

    x = related_beliefs_pcvec[:,pc_x]
    y = related_beliefs_pcvec[:,pc_y]
    
    dat = pd.DataFrame({'x':x, 'y':y, 'belief':related_beliefs})
    
    print("N samples:", len(x))
    return x, y, dat
```

```python
def draw_keyword_kde(keywords, pc_x=0, pc_y=1, palette='rocket', labeloff=0):
    
    color = palette
            
    x, y, dat = get_related_beliefs(keywords, pc_x, pc_y)
    g = sns.JointGrid(data=dat, x="x", y="y", space=0.0, ratio=7)
    
    
    sns.kdeplot(data=dat, x='x', y='y', fill=True, thresh=0, cmap=color, levels=100, ax=g.ax_joint)
    g.ax_joint.set_facecolor("black")
    g.plot_marginals(sns.histplot, color="#35294E", alpha=1, bins=25)
    #g.set_axis_labels(xlabel=None)
    
    
    ax=g.ax_joint
    ax.set_xticks([-20, 0, 20])
    ax.set_xlabel(f'PC{pc_x+1}')
    ax.set_ylabel(f'PC{pc_y+1}')
    
    if labeloff==0:
        if len(keywords) > 2:
            ax.text(.02, .97,  s=", ".join(keywords), ha='left', va='top', color='w', fontsize=22, transform=ax.transAxes)
        else:
            ax.text(.02, .97,  s=", ".join(keywords), ha='left', va='top', color='w', fontsize=23, transform=ax.transAxes)
        
    return g
```

```python
keywords = ['God']
draw_keyword_kde(keywords, labeloff=0)

#also test 'God' in PC2-3 and PC3-4. 
#Note that the pc_x, pc_y index is one less than the actual PC!
draw_keyword_kde(keywords, labeloff=0, pc_x=1, pc_y=2) #pc2 - pc3

draw_keyword_kde(keywords, labeloff=0, pc_x=2, pc_y=3) #pc3 - pc4
```

# User embedding 🙎‍♂️🙍‍♀️

## Calculating user embedding vector

 * Now let's represent users in the same belief embedding space using their prior beliefs. By averaging the belief vectors of a user one can caculate the user vector!

```python
#df_train = pd.read_pickle(f'../dataset-robust/full_data/df_train_idx0.p')
df_train.head()
```

```python
#fine-tuned model 
user2beliefs = {}
user2belief_vec = {}

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/belief_embedding_trainset_idx0.p', 'rb') as f:
    belief2vec = pickle.load(f)

for user, df_ in df_train.groupby('username'):
    user2beliefs[user] = df_['belief_statement'].values
    belief_vectors = np.array([belief2vec[b] for b in df_['belief_statement']])
    user2belief_vec[user] = np.mean(belief_vectors, axis=0)

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/user_embedding_trainset_idx0.p','wb')as f:
    pickle.dump(user2belief_vec, f)
```

```python
#base model 
user2beliefs_base = {}
user2belief_vec_base = {}

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults_basemodel/belief_embedding_trainset_idx0.p', 'rb') as f:
    belief2vec_base = pickle.load(f)

for user, df_ in df_train.groupby('username'):
    user2beliefs_base[user] = df_['belief_statement'].values
    belief_vectors_base = np.array([belief2vec_base[b] for b in df_['belief_statement']])
    user2belief_vec_base[user] = np.mean(belief_vectors_base, axis=0)

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults_basemodel/user_embedding_trainset_idx0.p','wb')as f:
    pickle.dump(user2belief_vec_base, f)
```

# Belief prediction task

## Preparation

 * In this section, we perform the belief prediction task, aiming to predict a user's belief (position) on a newly introduced debate. To achieve this, we leverage the user's position in the belief space and their distances from the two opposing beliefs under consideration.

```python
#generating two beliefs for a given debate title by adding a template phrase

def get_rephrase(s, pro=True):
    pro_phrase = 'I agree with the following: '
    con_phrase = 'I disagree with the following: '

    if pro:
        return pro_phrase + s
    else:
        return con_phrase + s
```

 * We need to generate all possible belief vectors (Pro vectors and Con vectors) for the beliefs in the test set.

```python
model = SentenceTransformer("Byunghwee/roberta_belief_finetuned")

belief2provec = {}
belief2convec = {}

df_test = pd.read_pickle(f'../dataset-robust/full_data/df_test_common_idx0.p') #test dataset which contains user data for individuals who appear in both training and test sets. 

title2provector = {}
title2convector = {}

for title in tqdm(df_test['debate_title'].unique()):
    title2provector[title] = model.encode(get_rephrase(title, pro=True))
    title2convector[title] = model.encode(get_rephrase(title, pro=False))
            
with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/test_title2provector0.p','wb')as f:
    pickle.dump(title2provector, f)

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/test_title2convector0.p','wb')as f:
    pickle.dump(title2convector, f)
```

 * For comparison, we find belief vectors also from the base model  

```python
basemodel = SentenceTransformer("roberta-base-nli-stsb-mean-tokens")

belief2provec = {}
belief2convec = {}

df_test = pd.read_pickle(f'../dataset-robust/full_data/df_test_common_idx0.p') #test dataset which contains user data for individuals who appear in both training and test sets. 

title2provector = {}
title2convector = {}

for title in tqdm(df_test['debate_title'].unique()):
    title2provector[title] = basemodel.encode(get_rephrase(title, pro=True))
    title2convector[title] = basemodel.encode(get_rephrase(title, pro=False))
            
with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/test_title2provector0.p','wb')as f:
    pickle.dump(title2provector, f)

with open(f'../dataset-robust/full_data/BeliefEmbeddingResults/test_title2convector0.p','wb')as f:
    pickle.dump(title2convector, f)
```

## Downstream task 

```python
def euclidean_accuracy(pro_vec, con_vec, user_vec):
    #Euclidian distance
    y_pro_euc = []
    y_con_euc = []
    d_min,d_max,d_avg,d_rel = [],[],[],[]
    
    for i in range(len(pro_vec)):
        a = np.linalg.norm(pro_vec[i] - user_vec[i])
        b = np.linalg.norm(con_vec[i] - user_vec[i])
        y_pro_euc.append(a)
        y_con_euc.append(b)
        
        d_min.append(np.min([a,b]))
        d_max.append(np.max([a,b]))
        d_avg.append(np.mean([a,b]))
        
        if not np.min([a,b]) == 0:
            d_rel.append( np.abs(a-b)/np.min([a,b]) )
        else:
            d_rel.append(0)
    
    y_pro_euc = np.array(y_pro_euc)
    y_con_euc = np.array(y_con_euc)
    y_hat = y_pro_euc - y_con_euc
    y_hat = list(map(lambda x: 'Pro' if x < 0 else 'Con', y_hat))

    return y_hat, d_min, d_max, d_avg, d_rel
```

#### Distribution of user history 

```python
def logbinning(data, numbin=30):
    # log-scaled bins
    bins = np.logspace(0, np.log2(max(data)), num=numbin, base=2)
    widths = (bins[1:] - bins[:-1])

    # Calculate histogram
    hist = np.histogram(data, bins=bins)
    # normalize by bin width
    hist_norm = hist[0]/widths

    # plot it!
    x = bins[:-1]
    y = hist_norm
    
    return x, y, widths
```

```python
result = pd.read_pickle(f"../dataset-robust/full_data/downstream_results/downstream_task_result_idx0.p") 

df_g = result.groupby('username')
  
votelengths = []
for g, df_ in df_g:
    votelengths.append(df_['user_length'].iloc[0])
    
x, y, widths = logbinning(votelengths, numbin=25)

fig, ax = plt.subplots(figsize=(5,4), dpi=150)
plt.loglog(x, y, 'o', ms=7, alpha=0.7, markeredgecolor='k', label='Data %d'%(i+1))
plt.xlabel(r'User history $L_v$', fontsize=15)
plt.ylabel('Probability density', fontsize=15)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.show()
```

### 2. Category (vs. Prediction accuracy)

 * We evaluate how different topics (categories) of debates influence the model's belief prediction performance. Are there certain categories of debates in which individuals' beliefs are more accurately predicted?

```python
with open('../dataset-robust/ddo_dataframe/debatekey2category_dict.p', 'rb') as f:
    debatekey2cat = pickle.load(f)
```

```python
from collections import defaultdict 

df_train = pd.read_pickle('../dataset-robust/full_data/df_train_common_idx0.p')

df_downstream = pd.read_pickle(f'../dataset-robust/full_data/downstream_results/downstream_task_result_idx0.p')

df_train['category'] = df_train['debate_key'].map(debatekey2cat)
df_downstream['category'] = df_downstream['debate_key'].map(debatekey2cat)

#number of debates with certain category in the "train set"
category2count = defaultdict(list) 

for cat, dat in df_train.groupby('category'):
    category2count[cat] = len(dat['debate_key'].unique())

df_downstream['cat_count_train'] = df_downstream['category'].map(category2count) 
df_downstream.to_pickle(f'../dataset-robust/full_data/downstream_results/downstream_task_result_idx0.p')
```

```python
#To ensure a sufficient number of samples for computing the mean accuracy, we select the top 20 categories that have more than 100 debates in the test set.
filtered_cat_list = df_downstream['category'].value_counts()
filtered_cat_list = filtered_cat_list[filtered_cat_list >= 100].index.to_list()
filtered_cat_list
```

```python
df_cat_f1 = pd.DataFrame(index=filtered_cat_list)


df_sample = pd.read_pickle(f'../dataset-robust/full_data/downstream_results/downstream_task_result_idx0.p')
df_sample = df_sample[df_sample['category'].isin(filtered_cat_list)]

category_labels = []
category_f1s = []

for cat, dat in df_sample.groupby('category'):
    category_f1s.append(f1_score(dat['position'], dat['predicted_outcome'], labels=['Pro','Con'], average='macro')) 
    category_labels.append(cat) 
        
sorted_indices = np.argsort(category_f1s)[::-1]
sorted_labels = [category_labels[i] for i in sorted_indices]
sorted_f1s = [category_f1s[i] for i in sorted_indices]
 
fig, ax = plt.subplots(figsize=(5.2,4), dpi=150)

plt.plot(np.arange(len(sorted_labels)), sorted_f1s, 's')
plt.xticks(np.arange(len(sorted_labels)), sorted_labels, rotation=90)
plt.axhline(0.5, ls='--', color='gray')
plt.show()
```
