### Check size of train & test triplet dataset

```python
dataset_path = '../dataset-robust/full_data/'

numtrip_train = []

for i in range(5):
    with open(dataset_path + 'train_triplet_idx%d.p'%i,'rb') as f:
        dat = pickle.load(f)
    print("train", i, len(dat))
    
    numtrip_train.append(len(dat))
    
numtrip_test_test = []
for i in range(5):
    with open(dataset_path + 'test_triplet_idx%d.p'%i,'rb') as f:
        dat = pickle.load(f)
    print("test-test", i, len(dat))
    
    numtrip_test_test.append(len(dat))    

print("average train/test-test triplet set size")
np.average(numtrip_train), np.average(numtrip_test_test)
```

## 2. Model fine-tuning

### Training Notes for Fine-Tuning S-BERT

- **Training Time Estimate**  
  - Fine-tuning **1 dataset** for **1 epoch** takes approximately **70 minutes** on an **Nvidia A100 GPU**.
  - Since we use **5-fold validation** (training 5 different models) and train each model for **3 epochs**, the total estimated training time is **15 ~ 20 hours**.

    - 5-Fold Validation Strategy: We use a 5-fold dataset split to train 5 separate models for validating experimental results. Each model is trained independently to ensure robust evaluation.

    - Epoch Limitations: Training beyond 3 epochs leads to a loss of semantic understanding, as evidenced by a significant drop in sentence similarity benchmark scores. To maintain optimal performance, the maximum number of epochs is set to 3.

### Downloading fine-tuned models

You can download the fine-tuned models in one of the following ways:

<1> manually downloading from the following link: 

1. Download __model_full_data.zip__ from the link: https://figshare.com/articles/dataset/Dataset_for_Neural_embedding_of_beliefs_reveals_the_role_of_relative_dissonance_in_human_decision-making_/28327019
2. Place the ZIP file inside the __'model'__ folder
3. Unzipping the file will generate __'full_data'__ folder inside the **'model'** folder.
4. The fine-tuned models will be located under **'../model/full_data/'** (e.g, 'model/full_data/roberta-base_idx0_epoch3/')

<2> By running the following code:
- it will automatically download and unzip neccessary model files

```python
import requests
import os
import zipfile
from tqdm import tqdm
from requests.adapters import HTTPAdapter

# Set paths
url = "https://figshare.com/ndownloader/files/52077533"

save_dir = "../model/"
zip_path = os.path.join(save_dir, "model_full_data.zip")
extract_path = "../model/"

# Ensure save directory exists
os.makedirs(save_dir, exist_ok=True)

# Configure requests session with optimized connection pooling
session = requests.Session()
adapter = HTTPAdapter(pool_connections=100, pool_maxsize=100) 
session.mount("https://", adapter)

# Download file with progress bar
print("🔽 Downloading ZIP file...")
response = session.get(url, stream=True)
total_size = int(response.headers.get("content-length", 0))

chunk_size = 1024 * 1024 

with open(zip_path, "wb") as file, tqdm(
    total=total_size, unit="B", unit_scale=True, desc="📥 Downloading"
) as bar:
    for chunk in response.iter_content(chunk_size=chunk_size):
        if chunk:
            file.write(chunk)
            file.flush() 
            bar.update(len(chunk))

print(f"✅ Download complete: {zip_path}")

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

#remove unnecessary files.
macosx_path = os.path.join(extract_path, "__MACOSX")
if os.path.exists(macosx_path):
    shutil.rmtree(macosx_path)  
    print("🗑️ Removed __MACOSX folder")

print(f"✅ Unzipped files to: {extract_path}")
```
